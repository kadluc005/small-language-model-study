```python
with open("../data/names.txt") as f:
    names = f.read().splitlines()

print(len(names))
print(names[:10])
```

```python
# prenom le plus court

smallest_name = names[0]
longest_name = names[0]
for name in names:
    if len(name) < len(smallest_name):
        smallest_name = name
    if len(name) > len(longest_name):
        longest_name = name    

print('smallest name:', '"'+smallest_name+'"', 'index:', names.index(smallest_name))
print('longest name:', '"'+longest_name+'"', 'index:', names.index(longest_name))
```
